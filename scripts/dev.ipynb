```python
roles_setup = [
    Werewolf(), Werewolf(), Werewolf(), Werewolf(),
    Villager(), Villager(), Villager(), Villager(),
    Seer(), Witch(), Hunter(), Idiot()
]
strategy_mapping = {
    RoleType.WEREWOLF: WerewolfBasicStrategy,
    RoleType.VILLAGER: VillagerBasicStrategy,
    RoleType.SEER: SeerBasicStrategy,
    RoleType.WITCH: WitchBasicStrategy,
    RoleType.HUNTER: HunterBasicStrategy,
    RoleType.IDIOT: IdiotBasicStrategy
}
config = GameConfig(
    roles_setup=roles_setup,
        strategy_mapping=strategy_mapping,
    win_condition=WinCondition.KILL_SIDE  # 屠边局
)
```

#### Election

```python
game.phase = GamePhase.DAY_ELECTION
```

```python
votes = {} # candidate -> vote count
candidates = []
alive_players = game.get_alive_players()
for p in alive_players:
    if p.elect_sheriff_strategy(game):
        candidates.append(p.player_id)
candidates
```

```python
alive_players[0].belief_state
```

```python
for pid in candidates:
    p = game.players[pid]
    speech, claims = p.speech_day_strategy(game)
    game.log(f"竞选者 玩家 {p.player_id} 发言: {speech}")
    for other_p in game.get_alive_players():
        if other_p.player_id != p.player_id:
            other_p.update_belief_after_speech(p.player_id, speech, claims, game)
```

```python
alive_players[0].belief_state
```

```python
alive_players[6].belief_state
```

```python
alive_players[9].belief_state
```

```python
for p in alive_players:
    if p.player_id not in candidates:
        vote = p.vote_sheriff(game, candidates)
        if vote in candidates:
            votes[vote] = votes.get(vote, 0) + 1
votes
```

```python
winner = max(votes, key=votes.get)
game.sheriff_id = winner
winner
```

```python
p = alive_players[8]
p.strategy.decide_vote_target(p, game)
```

```python
votes = {}
for p in game.get_alive_players():
    
    target = p.vote_day_strategy(game)
    print(p.role, target)
    if target and target in game.alive_players:
        weight = 1.5 if game.sheriff_id == p.player_id else 1.0
        votes[target] = votes.get(target, 0) + weight
votes
```

```python
if votes:
    exiled = max(votes, key=votes.get)
    exiled_player = game.players[exiled]
    
    # 白痴翻牌判定
    if exiled_player.role.role_type == RoleType.IDIOT and not exiled_player.is_idiot_revealed:
        exiled_player.is_idiot_revealed = True
        if game.sheriff_id == exiled:
            game.sheriff_id = None # 通常翻牌后也会失去警徽，视规则而定
    else:
        game.handle_death(exiled, is_vote=True)
```
